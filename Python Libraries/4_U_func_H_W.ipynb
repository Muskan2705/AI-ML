```python
#  1. add()
# arr1 = [10,20,30,40,50]
# arr2 = [5,4,3,2,1]

# add arrays 
# add 100 to arr1
# store result 
import numpy as np

arr1 = [10,20,30,40,50]
arr2 = [5,4,3,2,1]

result = np.add(arr1,arr2)
print("The sum of arr1 and arr2 :", result)


new_arr1 = np.add(arr1,100)
print("The sum of arr1 and 100 :", new_arr1)


final_result = np.add(new_arr1,arr2)
print("The final result after adding new_arr1 and arr2 :", final_result)
```

```python
# 2. subtract()
# arr1 = [100,80,60,40,20]
# arr2 = [10,20,30,40,50]

# - subtract arrays
# - subtract 25
# - identify negatives


arr1 = [100,80,60,40,20]
arr2 = [10,20,30,40,50]

subtract_result = np.subtract(arr1,arr2)
print("The result after subtracting arr2 from arr1 :", subtract_result)

subtract_25 = np.subtract(subtract_result,25)
print("The result after subtracting 25 from the previous result :", subtract_25)

negatives = subtract_25[subtract_25 < 0]
print("The negative values from the final result :", negatives)
```

```python
# 3. multiply()
# price = [50,100,150,200]
# qty = [2,4,1,3]

# total price
# price*18
# qty*2


price = [50,100,150,200]
qty = [2,4,1,3]

m = np.multiply(price,qty)
print("The total product is :", m)


new_price = np.multiply(price,18)
print("The new price after multiplying by 18 is :", new_price)

new_qty = np.multiply(qty,2)
print("The new quantity after multiplying by 2 is :", new_qty)
```

```python
# 4. divide()
# distance = [100,200,150,300]
# time = [2,4,3,6]

#speed
# distance/10
# speed/2


distance = [100,200,150,300]
time = [2,4,3,6]

speed = np.divide(distance,time)
print("The speed is :", speed)

new_distance = np.divide(distance,10)
print("The new distance after dividing by 10 is :", new_distance)

new_speed = np.divide(speed,2)
print("The new speed after dividing by 2 is :", new_speed)
```

```python
# 5. power()
# arr=[2,3,4,5]
# - Squares
# - Cubes
# - Power 4

arr=[2,3,4,5]
squares = np.power(arr,2)
print("The squares of the array are :", squares)

cubes = np.power(arr,3)
print("The cubes of the array are :", cubes)

power_4 = np.power(arr,4)
print("The power of 4 of the array are :", power_4)
```

```python
# 6. mod()
# arr=[15,22,39,48,57]
# - %2
# - %5
# - Odd numbers

arr=[15,22,39,48,57]
new_arr1 = np.mod(arr,2)
print("The result of new_arr1 : ", new_arr1)

new_arr2 = np.mod(arr,5)
print("The result of new_arr2 : ", new_arr2)

new_arr3 = np.mod(arr,2) ==1
print("The odd numbers in the array are :", new_arr3)
```

```python
# 7. remainder()
# a=[25,37,49,58], b=[4,5,6,7]
# - remainder(a,b)
# - remainder by 3

a=[25,37,49,58]
b=[4,5,6,7]
 
remainder_result = np.remainder(a,b)
print("The result is:", remainder_result)
```

```python
# 8. divmod()
# a=[15,28,39,47], b=[2,3,4,5]
# - Quotient
# - Remainder
# - Display separately

a=[15,28,39,47]
b=[2,3,4,5]

questient, remainder = np.divmod(a,b)
print(questient, remainder)

print("The quotient is:", questient)
print("The remainder is:", remainder)

```

```python
# 9. absolute()
# arr=[-15,-7,0,5,-20,30]
# - Positive values
# - Largest absolute


arr = [-15,-7,0,5,-20,30]

positive_values = np.absolute(arr)
print("The positive values are:", positive_values)


largest_absolute = np.max(arr)
print("The largest absolute value is:", largest_absolute)
```

```python
# 10. sqrt()
# arr=[1,4,9,16,25,36,49]
# - Square roots
# - Store
# - Sum

arr=[1,4,9,16,25,36,49]
sq_roots = np.sqrt(arr)
print("The square roots are:", sq_roots)

store = sq_roots
print("The stored square roots are:", store)

sum = np.sum(arr)
print("The sum of the square roots is:", sum)

sum1 = np.sum(sq_roots)
print("The sum of the square roots is:", sum1)


```

```python
# 11. log()
# arr=[1,2,4,8,16,32]
# - Natural log
# - Observe log(1)
# - log(arr*2)


arr=[1,2,4,8,16,32]
natural_log = np.log(arr)
print("The natural log of the array is:", natural_log)

observe_log1 = np.log(1)
print("The log of 1 is:", observe_log1)

log_arr = np.log(arr*2)
print("the log of an array is:",log_arr)


```

```python
# 12. exp()
# arr=[0,1,2,3,4]
# - exp
# - Compare exp(0), exp(1)
# - Round


arr=[0,1,2,3,4]
exp_result = np.exp(arr)
print("The exponential of the array is:", exp_result)

compare_result = np.exp(0), np.exp(1)
print("The comparison of exp(0) and exp(1) is:", compare_result)

round_result = np.round(exp_result)
print("The rounded exponential values are:", round_result)

```

COMPARISION AND REDUCTION

```python
# 13. maximum()
# marks1=[60,75,80,90,55], marks2=[65,70,85,88,60]
# - Max
# - Average of max

marks1=[60,75,80,90,55]
marks2=[65,70,85,88,60]

max_marks = np.max([marks1,marks2])
print("The maximum marks are:", max_marks)

avg_marks = np.average([max_marks])
print("The average of the maximum marks is:", avg_marks)

```

```python
# 14. minimum()
# Use same arrays
# - Min
# - Count <70

marks1=[60,75,80,90,55]
marks2=[65,70,85,88,60]

min_marks = np.min([marks1,marks2])
print("The minimum marks are:", min_marks)

all_marks = np.concatenate((marks1, marks2))
count = np.count_nonzero(all_marks < 70)
print("The count of marks less than 70 is:", count)

```
