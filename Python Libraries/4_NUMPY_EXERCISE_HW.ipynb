ARRAY JOINING (10 EXERCISE)

```python
# 1. Create two 1D arrays and join them

import numpy as np

arr1 = np.array([2,3,4])
arr2 = np.array([5,6,7])

a = np.concatenate((arr1, arr2))
print(a)
```

```python

# 2.	2. Join three arrays into one.

arr1 = np.array([2,3,4])
arr2 = np.array([5,6,7])
arr3 = np.array([8,9,10])

c = np.concatenate((arr1, arr2, arr3))
print(c)
```

```python
# 3.	3. Join two 2D arrays row-wise (axis=0).

arr1 = np.array([[2,3,4],[5,6,7]])
arr2 = np.array([[8,9,10],[11,12,13]])

b = np.concatenate((arr1, arr2), axis=0)
print(b)
```

```python
# 4.	4. Join two 2D arrays column-wise (axis=1).
arr1 = np.array([[2,3,4],[5,6,7]])
arr2 = np.array([[8,9,10],[11,12,13]])

d = np.concatenate((arr1, arr2), axis=1)
print(d)
```

```python
# 5.	5. Use np.stack() on two arrays.

arr1 = np.array([2,3,4])
arr2 = np.array([5,6,7])

s = np.stack((arr1, arr2))
print(s)
```

```python
# 6.	6. Use np.hstack() on two arrays

arr1 = np.array([2,3,4])
arr2 = np.array([5,6,7])

s = np.hstack((arr1, arr2))
print(s)
```

```python
# 7.	7. Use np.vstack() on two arrays.

arr1 = np.array([2,3,4])
arr2 = np.array([5,6,7])

s = np.vstack((arr1, arr2))
print(s)
```

```python
# 8.	8. Join arrays using np.concatenate().

arr1 = np.array([2,3,4])
arr2 = np.array([5,6,7])

c = np.concatenate((arr1, arr2))
print(c)
```

```python
# 9.	9. Predict the output before running concatenate.

arr1 = np.array([2,3,4])
arr2 = np.array([5,6,7])

print("the arr1 before concatenate:", arr1)
print("the arr2 before concatenate:", arr2)

c = np.concatenate((arr1, arr2))
print("the concatenated array:", c)
```

```python
# 10.	10. Create two student score arrays and merge them.

marks1= np.array([60,75,80,90,55])
marks2=np.array([65,70,85,88,60])

merged_marks = np.concatenate((marks1, marks2))
print("Merged student scores:", merged_marks)
```

```python
# 16.	6. Explain difference between split and array_split

# np.split() --it splits the array into equal parts 

# np.array_split() --it splits the array into equal or unequal parts


#split()
import numpy as np
arr = np.array([1, 2, 3, 4, 5, 6])
s = np.split(arr, 3)
print(s)


# array_split()
import numpy as np
arr = np.array([1, 2, 3, 4, 5, 6, 7])
s = np.array_split(arr, 3)
print(s)
```
