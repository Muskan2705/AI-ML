```python
# ufuncs -- universal functions

# vectorization -- process of converting iterative statements into vector-based operations 

# without universal function
# x = [1,2,3,4]
# y = [4,5,6,7]
# z = []

# for i,j in zip(x,y):
#     z.append(i+j)
#     print(z)


# with universal function 
import numpy as np
x = [1,2,3,4]
y = [4,5,6,7]
z = np.add(x,y)
print(z)
```

```python
# you can create your own universal function also by using -- frompyfunc() method 

import numpy as np
def myadd(x,y):
    return x+y

myadd = np.frompyfunc(myadd,2,1)
print(myadd([1,2,3,4],[5,6,7,8]))
```

```python
# ufuncs operation -- add(), subtract(), multiply(), divide(),power(), mod(), REMAINDER(),DIVMOD()

import numpy as np
arr1 = np.array([1,2,3,4])
arr2 = np.array([5,6,7,8])
new_arr = np.add(arr1,arr2)
print(new_arr)

arr1 = np.array([5,6,7,8])
arr2 = np.array([1,2,3,4])
new_arr = np.subtract(arr1,arr2)
print(new_arr)

arr1 = np.array([1,2,3,4])
arr2 = np.array([5,6,7,8])
new_arr = np.multiply(arr1,arr2)
print(new_arr)

arr1 = np.array([10,20,30,40])
arr2 = np.array([5,6,7,8])
new_arr = np.divide(arr1,arr2)
print(new_arr)
```

```python
import numpy as np
arr1 = np.array([10,20,30,40,50,60])
arr2 = np.array([3,5,6,8,2,33])
new_arr = np.power(arr1,arr2)
print(new_arr)
```

4_ AUGUST_NUMPY UNIVERSAL FUNCTION


```python
# sum 

import numpy as np
arr1 = np.array([2,8,125])
arr2 = np.array([3,3,15])
new_arr = np.sum([arr1,arr2])
print(new_arr)
```

```python
# summation over axis -- sum([a,b],axis= 1)

import numpy as np
arr1 = np.array([2,8,125])
arr2 = np.array([3,3,15])
new_arr = np.sum([arr1,arr2],axis=1)
print(new_arr)
```

```python
# cumulative summation -- cumsum()

import numpy as np
arr1 = np.array([2,8,125])
arr2 = np.array([3,3,15])
new_arr = np.cumsum([arr1,arr2])
print(new_arr)


```

```python
# PRODUCT OF ARRAYS

import numpy as np
arr1 = np.array([1,2,3,4])
arr2 = np.array([5,6,7,8])
x = np.prod([arr1,arr2])
print(x)
```

```python
# SUMATION PRODUCT
 
import numpy as np
arr1 = np.array([1,2,3,4])
arr2 = np.array([5,6,7,8])
x = np.prod([arr1,arr2],axis=1)
print(x)


# CUMULATIVE PRODUCT -- CUMPROD()

import numpy as np
arr1 = np.array([1,2,3,4])
arr2 = np.array([5,6,7,8])
x = np.cumprod([arr1,arr2])
print(x)
```

```python
# DISCREATE DIFFERENCE 
# SUBTRACTION BETWEEN ELEMENTS 
# [1,2,3,4] = [1-0,2-1,3-2,4-3] = [1,1,1,1]

import numpy as np
arr = np.array([10,45,43,23,60])
newarr = np.diff(arr)
print(newarr)
```

```python
# JOIN NUMPY ARRAYS

# putting the content of 2 or more arrays in a single array thru concatenate function concatenate()

import numpy as np
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
arr = np.concatenate((arr1,arr2))
print(arr)
```

```python
# class task -- join two 2-D arrays along axis 

import numpy as np
arr1 = np.array([[2,3,4],[5,6,7]])
arr2 = np.array([[9,7,6],[45,34,54]])
newarr = np.concatenate((arr1,arr2),axis=1)
print(newarr)


```

```python
# join array using stack function -- it is same as concatenate the only difference is that stacking is done along new axis 

# stack , hstack(), vstack(), dsatck().

import numpy as np
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
arr = np.stack((arr1,arr2),axis=1)
print(arr)

```

```python
# hstack()

import numpy as np
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
arr = np.hstack((arr1,arr2))
print(arr)

# vstack()

import numpy as np
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
arr = np.vstack((arr1,arr2))
print(arr)

# dstack()

import numpy as np
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
arr = np.dstack((arr1,arr2))
print(arr)

```

```python
# array split -- array splitting means breaking the array into multiple array



# array_split() method

import numpy as np
arr = np.array([1,2,3,4,5,6])
new_arr = np.array_split(arr,3)  # here 3 denotes, how many time we want array splitting - 3 times or 4 times 
print(new_arr)

```

```python
# class task -- split the 2-d array into 3 2D array 

import numpy as np
a = np.array([[23,45,67],[87,65,43],[34,56,78]])
new = np.array_split(a,3)
print(new)
```

```python
# class task -- split the 2D array containing 3 element each into 4 2D array

import numpy as np
a = np.array([[23,34,45],[45,67,10],[1,2,3],[4,5,6]])
new_arr = np.array_split(a,4)
print(new_arr)
```

```python

# Search array 
# we can search an array for a certain value and return the indexes that get a match, to search an array we use where() function 


import numpy as np 
arr = np.array([1,2,3,4,5,4,4,])
x = np.where(arr==4)
print(x)
```

```python
# class task -- find the indexes where the values are odd

import numpy as np 
arr = np.array([10,14,93,41,8,7])
x = np.where(arr%2 !=0)
print(x)

# for even 
import numpy as np 
arr = np.array([10,14,93,41,8,7])
x = np.where(arr%2 ==0)
print(x)
```

```python
# search sorted()

import numpy as np
arr = np.array([6,7,8,9])
v = np.searchsorted(arr,7)
print(v)
```
