```python
a = [1,2,3,4,5]
b = [10,12,13,16,17]

plt.plot(a,b,color='red',marker="*")
plt.title("line plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axix")
plt.grid(True)
plt.show()
```

```python
categories = ["Apple","lenovo","Samsung","Dell"]
value = [100,200,300,400]
plt.bar(categories,value,color=['Red','Green','Pink','Blue'],hatch="/")
plt.title("BAR PLOT")
plt.xlabel("CATEGORIES")
plt.ylabel("VALUES")
plt.show()
```

```python
study_hours = [1,2,3,4,5]
exam_scores = [10,12,13,16,17]

plt.scatter(study_hours,exam_scores,color="green")
plt.title("STUDY HOURS VS EXAM SCORES")
plt.xlabel("Study Hours")
plt.ylabel("Exam Scores")
plt.show()
```

```python
# we use this when we don't want to waste the side spaces, so we use this method 

fig,axs = plt.subplots(1,2,figsize=(10,4))
axs[0].bar(["Del","Mum","Hr"],[3,5,7])
axs[0].set_title("Bar")

axs[1].plot([1,2,3],[4,5,6])
axs[1].set_title("line")

plt.tight_layout()
plt.show()
```

```python
fig,axs = plt.subplots(1,4,figsize=(10,4))
axs[0].bar(["Del","Mum","Hr"],[3,5,7])
axs[0].set_title("BAR")

axs[1].plot([1,2,3],[4,5,6])
axs[1].set_title("LINE")

axs[2].scatter([1,2,3],[11,12,13])
axs[2].set_title("PATTERN")

axs[3].bar(["Apple","Mango","Banana"],[100,200,400])
axs[3].set_title('FRUITS')

plt.tight_layout()
plt.show()
```

```python
labels = ["Mon","Tue","Wed","Thur"]
size = [20,25,30,35]
plt.figure(figsize=(6,6))
plt.pie(size,labels = labels,autopct = "%1.1f%%",shadow=True,explode=[0.1,0.0,0.0,0.0]);
plt.title("DAY SALES")
plt.show()
```

```python

# SEABORN LIBRARY


import matplotlib as plt
import seaborn as sns
import pandas as pd
df = sns.load_dataset("tips")
```

```python
import matplotlib.pyplot as plt
sns.countplot(data=df,x="day")
plt.title("Count of Tips Per Day")
plt.show()
```

```python
sns.countplot(data=df,x='day',hue='sex')
plt.title('COUNTS OF TIPS PER DAY')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np





month = ['Jan','Feb','Mar','Apr','May','June','July','Aug','Sep','Oct','Nov','Dec']
sales_2024 = [50,55,53,60,70,75,72,68,70,74,80,74]
sales_2025=[]
for i in sales_2024:
    sales_2025.append(i+10)

plt.figure(figsize=(10,8))
plt.plot(month,sales_2024,label="sale 2024",color='blue',marker="o")
plt.plot(month,sales_2025,label="sale 2025",color='green',marker="s")
plt.xlabel("Month")
plt.ylabel("Monthly Sales Trend : 2024 vs 2025")
plt.grid("True")
plt.legend()
plt.show()
```
